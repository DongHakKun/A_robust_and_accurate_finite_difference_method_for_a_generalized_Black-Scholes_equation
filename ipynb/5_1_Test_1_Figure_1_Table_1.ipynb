## Figure 1 · 계산된 옵션 가격

$N=M=64$에서 $65\times65$개 격자값을 모두 계산한다. 그림의 시간축은 잔존만기 대신 원래 좌표 $t\in[0,T]$이다. 원문의 표면도와 같은 수치 설정을 사용하며, strike 주변 시간 단면을 추가로 그려 가격 변화를 읽는다.

```c++
BlackScholesSolver figure(64, 64, 0.04, 0.06);
figure.solveTest1();
printf("N=%d, M=%d, epsilon=%.1e, alpha=%.2f, beta*=%.2f\n",
       figure.N, figure.M, figure.epsilon, figure.alpha, figure.betaStar);
printf("U(K,0)=%.10f, minimum U=%.8e\n", figure.value(figure.N/4,0), ErrorAnalysis::minimumValue(figure));
printf("j=0: max interior off-diagonal=%.8e, min dominance margin=%.8e\n",
       figure.largestInteriorOffDiagonal(), figure.minimumDiagonalMargin());
plotSolution(figure, "test1", "Test 1");
displayPNG("../results/test1_surface.png");
```

## 기준해 · $U^{2048,1024}$

논문과 동일하게 $N_{\rm ref}=2048$, $M=1024$인 수치해를 참조해로 사용한다. 전체 시간 이력을 보관하며, coarse grid와 같은 시간층 $t_j$에서 **공간 선형 보간**한다.

$$\bar U(x,t_j)=\frac{x_{k+1}^{\rm ref}-x}{x_{k+1}^{\rm ref}-x_k^{\rm ref}}U_k^{j,\rm ref}
+\frac{x-x_k^{\rm ref}}{x_{k+1}^{\rm ref}-x_k^{\rm ref}}U_{k+1}^{j,\rm ref}.$$

```c++
const int M = 1024;
const int Nref = 2048;
auto referenceStart = std::chrono::steady_clock::now();
BlackScholesSolver reference(Nref, M, 0.04, 0.06);
reference.solveTest1();
double referenceSeconds = std::chrono::duration<double>(std::chrono::steady_clock::now()-referenceStart).count();
printf("Reference: N=%d, M=%d, %.3f seconds, U history %.2f MiB\n",
       Nref, M, referenceSeconds, (Nref+1.0)*(M+1.0)*sizeof(double)/(1024.0*1024.0));
```

## Table 1 · 공간 격자 세분

$$e^{N,M}=\max_{i,j}|U_i^j-\bar U(x_i,t_j)|,\qquad
\bar r^{N,M}=\log_2\left(\frac{e^{N,M}}{e^{2N,M}}\right).$$

$M=1024$를 고정하고 $N=128,256,512,1024$를 계산한다. 전체 시공간 격자에서 최대오차와 그 위치 $(x_i,t_j)$를 기록한다. 아래 `paperErrors`는 원문에 인쇄된 비교용 값이며, `errors`는 C++ 풀이와 보간으로 새로 계산한다.

```c++
const int count = 4;
int nValues[count] = {128, 256, 512, 1024};
double paperErrors[count] = {6.8027e-02, 3.1258e-02, 1.2588e-02, 3.4189e-03};
double errors[count];
double errorX[count];
double errorTime[count];
for(int k=0; k<count; k++)
{
    BlackScholesSolver coarse(nValues[k], M, 0.04, 0.06);
    coarse.solveTest1();
    int iMax = 0;
    int jMax = 0;
    errors[k] = ErrorAnalysis::maxError(coarse, reference, &iMax, &jMax);
    errorX[k] = coarse.x[iMax];
    errorTime[k] = jMax*coarse.tau;
}

printf("   M     N       computed e      paper e     rate      x at max     t at max\n");
FILE* errorFile = openOutput("../results/test1_errors.csv");
fprintf(errorFile, "N,error,rate,paper_error,x_at_max,t_at_max\n");
for(int k=0; k<count; k++)
{
    printf("%4d %5d   %14.8e  %11.4e  ", M, nValues[k], errors[k], paperErrors[k]);
    fprintf(errorFile, "%d,%.17g,", nValues[k], errors[k]);
    if(k+1<count)
    {
        double rate = log(errors[k]/errors[k+1])/log(2.0);
        printf("%7.3f", rate);
        fprintf(errorFile, "%.17g", rate);
    }
    else printf("%7s", "-");
    printf("   %11.6f   %10.7f\n", errorX[k], errorTime[k]);
    fprintf(errorFile, ",%.17g,%.17g,%.17g\n", paperErrors[k], errorX[k], errorTime[k]);
}
fclose(errorFile);
```

### 원문 수치와 비교

원문 Table 1의 수렴률은 $1.122,\ 1.312,\ 1.880$이다. 인쇄된 오차는 유효숫자 다섯 자리이므로 아래에서는 상대 차이와 **인쇄 반올림 단위 안에 들어오는지**를 함께 확인한다.

```c++
printf("    N   relative difference to printed value      rounded match\n");
for(int k=0; k<count; k++)
{
    double halfUnit = 0.5*pow(10.0, floor(log10(paperErrors[k]))-4.0);
    double relative = fabs(errors[k]-paperErrors[k])/paperErrors[k];
    bool matches = fabs(errors[k]-paperErrors[k]) <= halfUnit + 2e-10;
    printf("%5d                  %12.4e                 %s\n",
           nValues[k], relative, matches ? "yes" : "NO");
}
```

- $N=512\to1024$에서 관측 수렴률은 약 **1.880**이다. 공간 세분에 따라 오차가 줄어드는 양상을 Theorem 1의 공간 2차 수렴률과 비교한다.

## 추가 실험 · $\varepsilon=1/2$

$M=1024$, $N_{\rm ref}=2048$에서 coarse와 참조해 모두 $\varepsilon=1/2$로 계산한다. `p order (기존)`은 앞의 $\varepsilon=10^{-4}$ 실험의 관측률이다.

```c++
{
    const double epsilonHalf = 1.0 / 2.0;
    BlackScholesSolver halfReference(Nref, M, 0.04, 0.06, epsilonHalf);
    halfReference.solveTest1();
    double halfErrors[count];
    for(int k=0; k<count; k++)
    {
        BlackScholesSolver coarse(nValues[k], M, 0.04, 0.06, epsilonHalf);
        coarse.solveTest1();
        halfErrors[k] = ErrorAnalysis::maxError(coarse, halfReference);
    }

    printf("   M     N       computed e    p order    p order (기존)\n");
    FILE* halfFile = openOutput("../results/test1_epsilon_half_errors.csv");
    fprintf(halfFile, "M,N,computed e,p order,p order (기존)\n");
    for(int k=0; k<count; k++)
    {
        printf("%4d %5d   %14.8e  ", M, nValues[k], halfErrors[k]);
        fprintf(halfFile, "%d,%d,%.17g,", M, nValues[k], halfErrors[k]);
        if(k+1<count)
        {
            double order = log(halfErrors[k]/halfErrors[k+1])/log(2.0);
            double originalOrder = log(errors[k]/errors[k+1])/log(2.0);
            printf("%9.3f %17.3f\n", order, originalOrder);
            fprintf(halfFile, "%.17g,%.17g\n", order, originalOrder);
        }
        else
        {
            printf("%9s %17s\n", "-", "-");
            fprintf(halfFile, ",\n");
        }
    }
    fclose(halfFile);
}
```
